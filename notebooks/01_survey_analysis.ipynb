## Xếp hạng mô tả các mục Likert

Điểm trung bình chỉ là mô tả trên thang 1–5. Cần xem cùng trung vị và phân phối; không diễn giải như quan hệ nhân quả.

```python
with (ROOT/'reports/likert_summary.csv').open(encoding='utf-8-sig',newline='') as f:
    items=list(csv.DictReader(f))
for r in sorted(items,key=lambda x:float(x['mean']),reverse=True):
    print(f"Q{r['question']}: mean={float(r['mean']):.2f}; median={r['median']}; top2={float(r['top2_pct']):.1f}%")
checks=json.loads((ROOT/'reports/quality.json').read_text(encoding='utf-8'))
assert all(c['pass'] for c in checks)
print(len(checks),'checks PASS')
```
